```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### Load the dataset


```python
df = pd.read_csv('Customers_Fakedata.csv')
df.head()
```

```python
print("=== Dataset Shape ===")
print(df.shape)

print("\n=== First 5 Rows ===")
display(df.head())

print("\n=== Dataset Information ===")
df.info()
```

```python
print("=== Column Names ===")

for i, col in enumerate(df.columns):
    print(i, repr(col))
```

### Missing Values

```python
print("=== Missing Values Count ===")
print(df.isna().sum())

print("\n=== Missing Values Percentage ===")

missing_percentage = (df.isna().sum() / len(df)) * 100

print(missing_percentage.round(2))
```

### Duplicate Rows


```python
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

# Data Cleaning

```python
print("=== Gender Values ===")
print(df['Gender'].value_counts(dropna=False))
```

```python
print("=== Age Statistics ===")
print(df['Age'].describe())

print("\n=== Smallest 10 Ages ===")
print(df['Age'].nsmallest(10))

print("\n=== Largest 10 Ages ===")
print(df['Age'].nlargest(10))
```

```python
print("=== Purchase Amount Statistics ===")
print(df['PurchaseAmount'].describe())

print("\n=== Missing Purchase Amount ===")
print(df['PurchaseAmount'].isna().sum())
```

```python
print("=== Rating Values ===")

print(
    df['Rating']
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
print("=== Product Category Values ===")

print(
    df['ProductCategory']
    .value_counts(dropna=False)
)
```

```python
print("=== Purchase Date Sample ===")
print(df['PurchaseDate'].head(10))

print("\n=== Purchase Date Data Type ===")
print(df['PurchaseDate'].dtype)
```

```python
print("=== Removing Unnecessary Columns ===")

columns_to_drop = [
    'Unnamed',
    '  Gender  '
]

df = df.drop(
    columns=[col for col in columns_to_drop if col in df.columns]
)

print("\n=== Remaining Columns ===")
print(df.columns.tolist())
```

```python
print("=== Gender Before Cleaning ===")

print(df['Gender'].value_counts(dropna=False))


df['Gender'] = df['Gender'].str.strip()


df['Gender'] = df['Gender'].str.title()


df['Gender'] = df['Gender'].replace({
    'M': 'Male',
    'F': 'Female'
})


print("\n=== Gender After Cleaning ===")

print(df['Gender'].value_counts(dropna=False))
```

```python
print("=== Age Before Cleaning ===")

print(df['Age'].describe())

invalid_age_mask = (
    (df['Age'] < 15) |
    (df['Age'] > 100)
)

print("\n=== Invalid Age Count ===")
print(invalid_age_mask.sum())

df.loc[invalid_age_mask, 'Age'] = np.nan


print("\n=== Missing Age After Cleaning ===")
print(df['Age'].isna().sum())


print("\n=== Age After Cleaning ===")
print(df['Age'].describe())
```

```python
print("=== Rating Before Cleaning ===")

print(df['Rating'].value_counts(dropna=False).sort_index())


invalid_rating_mask = (
    (df['Rating'] < 1) |
    (df['Rating'] > 5)
)

print("\n=== Invalid Rating Count ===")
print(invalid_rating_mask.sum())

df.loc[invalid_rating_mask, 'Rating'] = np.nan

print("\n=== Rating After Cleaning ===")

print(df['Rating'].value_counts(dropna=False).sort_index())
```

```python
print("=== Purchase Date Before Cleaning ===")

print(df['PurchaseDate'].head(10))

df['PurchaseDate'] = pd.to_datetime(
    df['PurchaseDate'],
    format='%m/%d/%Y',
    errors='coerce'
)


print("\n=== Invalid Purchase Dates ===")

print(df['PurchaseDate'].isna().sum())


print("\n=== Purchase Date Data Type ===")

print(df['PurchaseDate'].dtype)
```

## Handle  Missing Values

```python
print("=== Handling Categorical Missing Values ===")

df['Gender'] = df['Gender'].fillna('Unknown')

df['ProductCategory'] = df['ProductCategory'].fillna('Unknown')

df['Phone'] = df['Phone'].fillna('Unknown')


print("\n=== Missing Values After Categorical Cleaning ===")

print(df[['Gender', 'ProductCategory', 'Phone']].isna().sum())
```

### Handle Duplicates

```python
print("=== Duplicate Rows Before Removal ===")

duplicate_count = df.duplicated().sum()

print(duplicate_count)


df = df.drop_duplicates()


print("\n=== Duplicate Rows After Removal ===")

print(df.duplicated().sum())
```

### Final Validation


```python
print("=== FINAL DATA QUALITY CHECK ===")


print("\nDataset Shape:")
print(df.shape)


print("\nMissing Values:")
print(df.isna().sum())


print("\nMissing Values Percentage:")
print(
    ((df.isna().sum() / len(df)) * 100)
    .round(2)
)


print("\nDuplicate Rows:")
print(df.duplicated().sum())


print("\nData Types:")
print(df.dtypes)
```

### Save  Clean Dataset

```python
df.to_csv(
    'Customers_Cleaned.csv',
    index=False
)

print("=== Cleaning Completed Successfully ===")
print("Saved file: Customers_Cleaned.csv")
```

Product Category Analysis

```python
category_analysis = df.groupby('ProductCategory').agg(
    Total_Sales=('PurchaseAmount', 'sum'),
    Purchase_Count=('PurchaseAmount', 'count'),
    Average_Purchase=('PurchaseAmount', 'mean')
).sort_values(
    by='Total_Sales',
    ascending=False
)

print(category_analysis)
```

Sales by Time

```python
print(f"First Purchase Date: {df['PurchaseDate'].min()}")
print(f"Last Purchase Date: {df['PurchaseDate'].max()}")

yearly_sales = df.groupby(
    df['PurchaseDate'].dt.year
)['PurchaseAmount'].sum()

print("\n=== Sales by Year ===")
print(yearly_sales)
```
